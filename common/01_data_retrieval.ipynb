# 01 — Yield Curve Data Retrieval (5 Countries)

Download daily yield curve data:
- **UK**: Bank of England nominal spot curve
- **US**: FRED Treasury constant-maturity rates
- **EU**: ECB AAA-rated government bond spot rates
- **CA**: Bank of Canada benchmark bond yields
- **BR**: ANBIMA/IPEA Brazilian DI pre-fixed curve

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_retrieval import (
    fetch_uk_yields, fetch_us_yields, fetch_eu_yields,
    fetch_ca_yields, fetch_br_yields, align_yield_data,
)

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14, 5)

START = '2005-01-01'
END = '2026-04-30'
```

## 1. Fetch All Countries

```python
uk = fetch_uk_yields(start_date=START, end_date=END)
print(f'UK: {uk.shape}')

us = fetch_us_yields(start_date=START, end_date=END)
print(f'US: {us.shape}')

eu = fetch_eu_yields(start_date=START, end_date=END)
print(f'EU: {eu.shape}')

ca = fetch_ca_yields(start_date=START, end_date=END)
print(f'CA: {ca.shape}')

br = fetch_br_yields(start_date=START, end_date=END)
print(f'BR: {br.shape}')
```

## 2. Align

```python
combined = align_yield_data(uk, us, eu, ca, br)
print(f'Combined: {combined.shape}')
print(f'Range: {combined.index.min()} to {combined.index.max()}')
print(f'NaN: {combined.isna().sum().sum()}')
combined.describe()
```

## 3. Visualization — 10Y Yields

```python
fig, ax = plt.subplots(figsize=(14, 6))
for col, label in [('UK_10Y','UK'), ('US_10Y','US'), ('EU_10Y','EU'),
                    ('CA_10Y','CA'), ('BR_12Y','BR (12Y)')]:
    if col in combined.columns:
        ax.plot(combined.index, combined[col], label=label, alpha=0.8, lw=0.8)
ax.set_ylabel('Yield (%)')
ax.set_title('Long-Term Government Bond Yields (5 Countries)')
ax.legend()
plt.tight_layout()
plt.show()
```

## 4. Save

```python
combined.to_csv('../data/raw/aligned_yields_5c.csv')
print('Saved aligned_yields_5c.csv')
```
